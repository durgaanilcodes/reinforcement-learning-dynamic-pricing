```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.boxplot(df["adr"])

plt.title("ADR Outlier Detection")

plt.ylabel("Average Daily Rate")

plt.show()
```

## Observation

The detected outliers represent genuine booking variations and were retained because they may reflect real-world pricing behavior.
