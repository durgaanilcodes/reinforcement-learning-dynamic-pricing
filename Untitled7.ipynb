```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
states = 100
actions = 3

Q = np.zeros((states, actions))

print(Q.shape)
```

```python
learning_rate = 0.1
discount_factor = 0.9
epsilon = 0.1

episodes = 500
```

```python
rewards = []
```

```python
for episode in range(episodes):

    state = np.random.randint(0,100)

    total_reward = 0

    for step in range(20):

        if np.random.rand() < epsilon:
            action = np.random.randint(0,3)
        else:
            action = np.argmax(Q[state])

        next_state = np.random.randint(0,100)

        reward = np.random.randint(1,10)

        Q[state,action] = Q[state,action] + learning_rate * (
            reward +
            discount_factor * np.max(Q[next_state])
            - Q[state,action]
        )

        state = next_state

        total_reward += reward

    rewards.append(total_reward)
```

```python
print("Training Completed")
```

```python
plt.figure(figsize=(10,5))

plt.plot(rewards)

plt.title("Reward per Episode")

plt.xlabel("Episode")

plt.ylabel("Reward")

plt.grid()

plt.show()
```

```python
average_reward = np.mean(rewards)

print("Average Reward :", average_reward)
```

```python
best_action = np.argmax(Q,axis=1)

print(best_action[:20])
```

```python
plt.figure(figsize=(8,5))

plt.hist(best_action)

plt.title("Best Action Distribution")

plt.xlabel("Action")

plt.ylabel("Frequency")

plt.show()
```
