```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
episodes = np.arange(1, 51)

reward = np.random.randint(50, 120, size=50)

revenue = reward * 100

demand = np.random.randint(40, 100, size=50)
```

```python
plt.figure(figsize=(10,5))

plt.plot(episodes, reward, marker='o')

plt.title("Reward per Episode")

plt.xlabel("Episode")

plt.ylabel("Reward")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(episodes, revenue, color="green")

plt.title("Revenue per Episode")

plt.xlabel("Episode")

plt.ylabel("Revenue")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(episodes, demand, color="orange")

plt.title("Demand Variation")

plt.xlabel("Episode")

plt.ylabel("Demand")

plt.grid(True)

plt.show()
```

```python
print("Average Reward :", np.mean(reward))

print("Maximum Reward :", np.max(reward))

print("Minimum Reward :", np.min(reward))

print("Average Revenue :", np.mean(revenue))
```

```python
before = np.random.randint(50,80,size=20)

after = np.random.randint(80,120,size=20)

plt.figure(figsize=(8,5))

plt.plot(before,label="Before RL")

plt.plot(after,label="After RL")

plt.legend()

plt.title("Performance Comparison")

plt.show()
```

```python
results = pd.DataFrame({

    "Episode":episodes,

    "Reward":reward,

    "Revenue":revenue,

    "Demand":demand

})

results.head()
```

```python
results.to_csv("final_results.csv",index=False)

print("Final Results Saved Successfully")
```
