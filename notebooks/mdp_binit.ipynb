## Introduction

Markov Decision Process (MDP) is a mathematical framework used in Reinforcement Learning to model sequential decision-making. It helps an agent choose the best action in a given state to maximize long-term rewards.

## What is Markov Decision Process?

A Markov Decision Process (MDP) is a model that represents decision-making situations where outcomes depend on the current state, the chosen action, and a reward received from the environment.

MDP is widely used in Reinforcement Learning because it provides a structured way to model interactions between an agent and its environment.

## Components of MDP

An MDP consists of five main components:

1. State (S)
2. Action (A)
3. Reward (R)
4. Policy (π)
5. Transition Probability (P)

## 1. State (S)

A state represents the current situation of the environment.

Example:

Hotel Rooms Available = 20

Days Remaining = 15

State = (20,15)

## 2. Action (A)

An action is the decision taken by the agent.

Examples:

- Increase room price
- Decrease room price
- Keep the same price

## 3. Reward (R)

A reward is the feedback received after taking an action.

Example:

If a hotel room is sold for ₹5000,

Reward = ₹5000

If no room is sold,

Reward = ₹0

## 4. Policy (π)

A policy is the strategy followed by the agent to decide which action to take in every state.

The objective is to maximize the total reward over time.

## 5. Transition Probability (P)

Transition Probability defines the chance of moving from one state to another after performing an action.

Example:

Current State

↓

Increase Price

↓

Customer buys the room

↓

New State

# Markov Property

The Markov Property states that the future depends only on the current state and action, not on the sequence of previous states.

Example:

Rooms Left = 5

Days Left = 2

The pricing decision depends only on the current situation, not on what happened during previous days.
