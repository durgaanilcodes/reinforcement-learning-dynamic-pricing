```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
np.random.seed(42)
```

```python
num_records = 1000
```

```python
synthetic_df = pd.DataFrame({
    "Booking_ID": range(1, num_records + 1),

    "Hotel_Type": np.random.choice(
        ["City Hotel", "Resort Hotel"],
        num_records
    ),

    "Season": np.random.choice(
        ["Peak", "Off-Peak", "Holiday"],
        num_records
    ),

    "Rooms_Available": np.random.randint(
        10, 200, num_records
    ),

    "Base_Price": np.random.randint(
        50, 300, num_records
    ),

    "Demand_Level": np.random.choice(
        ["Low", "Medium", "High"],
        num_records
    ),

    "Booking_Probability": np.round(
        np.random.uniform(0.2, 1.0, num_records),
        2
    )
})
```

```python
price_multiplier = np.random.uniform(0.8, 1.5, num_records)

synthetic_df["Dynamic_Price"] = (
    synthetic_df["Base_Price"] * price_multiplier
).round(2)

synthetic_df.head()
```

```python
synthetic_df["Revenue"] = (
    synthetic_df["Dynamic_Price"] *
    synthetic_df["Booking_Probability"]
).round(2)
```
