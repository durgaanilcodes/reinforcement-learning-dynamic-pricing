```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
real_df = pd.read_csv("cleaned_hotel_bookings.csv")
synthetic_df = pd.read_csv("synthetic_booking_data.csv")

print("Datasets loaded successfully!")
```

```python
print("Real Dataset Information")
real_df.info()

print("\nSynthetic Dataset Information")
synthetic_df.info()
```

## Observation

The dataset information confirms the number of records, columns, data types, and non-null values for both datasets. This helps verify that the datasets have been loaded correctly.

```python
print("Real Dataset Shape:", real_df.shape)
print("Synthetic Dataset Shape:", synthetic_df.shape)
```

```python
print("Missing Values in Real Dataset")
print(real_df.isnull().sum())

print("\nMissing Values in Synthetic Dataset")
print(synthetic_df.isnull().sum())
```

## Observation

Both datasets were checked for missing values. A valid training dataset should not contain missing values in important features.

```python
print("Duplicate Records in Real Dataset:", real_df.duplicated().sum())

print("Duplicate Records in Synthetic Dataset:", synthetic_df.duplicated().sum())
```

## Observation

Duplicate records can introduce bias during model training. The datasets were verified to ensure duplicate entries were removed.


```python
print("Real Dataset Data Types")
print(real_df.dtypes)

print("\nSynthetic Dataset Data Types")
print(synthetic_df.dtypes)
```
