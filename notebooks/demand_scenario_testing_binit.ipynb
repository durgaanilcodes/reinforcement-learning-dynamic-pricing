```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
scenario_df = pd.DataFrame({

    "Scenario":[
        "Low Demand",
        "Medium Demand",
        "High Demand",
        "Peak Season",
        "Off Season"
    ],

    "Rooms_Available":[
        180,
        120,
        50,
        30,
        190
    ],

    "Booking_Probability":[
        0.30,
        0.60,
        0.90,
        0.95,
        0.20
    ],

    "Base_Price":[
        100,
        120,
        150,
        180,
        90
    ]
})

scenario_df
```

```python
scenario_df["Dynamic_Price"] = [
    90,
    120,
    180,
    220,
    80
]

scenario_df
```

```python
scenario_df["Expected_Revenue"] = (
    scenario_df["Dynamic_Price"] *
    scenario_df["Booking_Probability"]
).round(2)

scenario_df
```

```python
scenario_df.to_csv(
    "demand_scenarios.csv",
    index=False
)

print("Demand scenario dataset saved successfully!")
```

```python
scenario_df.info()
```

## Observation

The demand scenario dataset contains five different booking situations representing various demand levels used to test the pricing environment.

```python
plt.figure(figsize=(8,5))

plt.bar(
    scenario_df["Scenario"],
    scenario_df["Dynamic_Price"]
)

plt.title("Dynamic Price under Different Demand Scenarios")

plt.xlabel("Demand Scenario")

plt.ylabel("Price")

plt.xticks(rotation=20)

plt.show()
```

## Observation

Dynamic prices increase during high-demand periods and decrease during low-demand periods, demonstrating the expected pricing behavior.

```python
plt.figure(figsize=(8,5))

plt.bar(
    scenario_df["Scenario"],
    scenario_df["Expected_Revenue"]
)

plt.title("Expected Revenue by Demand Scenario")

plt.xlabel("Demand Scenario")

plt.ylabel("Revenue")

plt.xticks(rotation=20)

plt.show()
```

## Observation

Expected revenue is highest during peak season and high-demand scenarios because both booking probability and price are higher.

```python
plt.figure(figsize=(8,5))

plt.plot(
    scenario_df["Scenario"],
    scenario_df["Booking_Probability"],
    marker="o"
)

plt.title("Booking Probability Across Demand Scenarios")

plt.xlabel("Demand Scenario")

plt.ylabel("Booking Probability")

plt.grid(True)

plt.show()
```

## Observation

Booking probability increases from low-demand scenarios to peak-season scenarios, which aligns with expected customer behavior.

# Environment Testing Results

The pricing environment was tested under five demand scenarios.

The following behaviours were verified:

- Low demand → Lower pricing
- Medium demand → Standard pricing
- High demand → Increased pricing
- Peak season → Maximum pricing
- Off season → Discounted pricing

The pricing logic responded appropriately to changes in demand.
