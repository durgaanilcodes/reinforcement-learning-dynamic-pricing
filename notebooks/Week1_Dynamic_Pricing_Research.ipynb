## 1. Introduction

Dynamic pricing is a pricing strategy where the price of a product or service changes based on various factors such as customer demand, remaining inventory, time, season, and competitor prices.

In the travel and hospitality industry, airlines and hotels use dynamic pricing to maximize revenue while ensuring that seats and rooms are sold before departure or the booking date.

## 2. What is Dynamic Pricing?

Dynamic pricing is an intelligent pricing strategy where prices are continuously adjusted according to market conditions.

### Factors affecting dynamic pricing

- Customer demand
- Remaining inventory
- Days left before departure/check-in
- Seasonal demand
- Competitor prices
- Special events and holidays

### Benefits

- Maximizes revenue
- Improves occupancy
- Reduces unsold inventory
- Responds to market demand

## 3. Dynamic Pricing in Airlines

Airlines use dynamic pricing to maximize ticket revenue.

Factors influencing ticket prices:

- Number of seats available
- Days remaining before departure
- Route popularity
- Passenger demand
- Holiday seasons
- Competitor pricing

Example:
A flight ticket booked 30 days before departure is usually cheaper than the same ticket booked one day before departure.

## 4. Dynamic Pricing in Hotels

Hotels frequently change room prices depending on demand.

Factors include:

- Room availability
- Weekend bookings
- Festival seasons
- Local events
- Customer demand
- Competitor hotel prices

Dynamic pricing helps hotels maximize occupancy and revenue.

# 8. Visualizations

The following visualizations use **sample (simulated) data** to illustrate how dynamic pricing concepts work. They are intended for understanding pricing behavior and are not based on actual airline or hotel booking data.

```python
import matplotlib.pyplot as plt
import numpy as np
```

## Visualization 1: Demand vs Price

This graph shows that customer demand decreases as the ticket price increases.

```python
price = [50, 75, 100, 125, 150, 175, 200]
demand = [100, 92, 80, 65, 48, 30, 18]
plt.figure(figsize=(8,5))
plt.plot(price, demand, marker='o')
plt.title("Demand vs Price")
plt.xlabel("Price ($)")
plt.ylabel("Customer Demand")
plt.grid(True)
plt.savefig("Demand_vs_Price.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

As prices increase, customer demand decreases. This inverse relationship is a key principle of dynamic pricing.

## Visualization 2: Revenue vs Price

```python
price = [50,75,100,125,150,175,200]
revenue = [5000,6900,8000,8100,7200,5200,3600]
plt.figure(figsize=(8,5))
plt.plot(price,revenue,marker='o')
plt.title("Revenue vs Price")
plt.xlabel("Price ($)")
plt.ylabel("Revenue ($)")
plt.grid(True)
plt.savefig("Revenue_vs_Price.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

Revenue increases up to an optimal price point and then decreases because higher prices reduce customer demand.

## Visualization 3: Booking Probability vs Price

```python
price = [50,75,100,125,150,175,200]
probability = [0.95,0.88,0.75,0.58,0.42,0.25,0.10]
plt.figure(figsize=(8,5))
plt.plot(price,probability,marker='o')
plt.title("Booking Probability vs Price")
plt.xlabel("Price ($)")
plt.ylabel("Booking Probability")
plt.grid(True)
plt.savefig("Booking_Probability.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

As prices rise, customers become less likely to complete a booking.

## Visualization 4: Remaining Seats Over Time

```python
days = [30,25,20,15,10,5,0]
seats = [120,105,88,65,40,18,0]
plt.figure(figsize=(8,5))
plt.plot(days,seats,marker='o')
plt.gca().invert_xaxis()
plt.title("Remaining Seats Over Time")
plt.xlabel("Days Before Departure")
plt.ylabel("Available Seats")
plt.grid(True)
plt.savefig("Remaining_Seats.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

As the departure date approaches, available seats decrease due to bookings.

## Visualization 5: Dynamic Ticket Price Over Time

```python
days = [30,25,20,15,10,5,0]
price = [120,125,135,150,145,110,80]
plt.figure(figsize=(8,5))
plt.plot(days,price,marker='o')
plt.gca().invert_xaxis()
plt.title("Dynamic Ticket Price Over Time")
plt.xlabel("Days Before Departure")
plt.ylabel("Ticket Price ($)")
plt.grid(True)
plt.savefig("Dynamic_Price.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

Prices may increase as demand grows and then decrease near departure to encourage the sale of remaining seats.
