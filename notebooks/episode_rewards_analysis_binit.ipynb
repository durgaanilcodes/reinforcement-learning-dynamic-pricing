```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("episode_rewards.csv")

print("Dataset loaded successfully!")
```

```python
df.head(10)
```

```python
df.info()
```

## Observation

The dataset contains two columns:

- Episode
- Reward

Each row represents the reward obtained by the Reinforcement Learning agent during one training episode.

```python
df.describe()
```

## Observation

The statistical summary provides information about the reward distribution, including average reward, minimum reward, maximum reward, and standard deviation.

```python
df.isnull().sum()
```

## Observation

The dataset does not contain missing values, ensuring that all episode rewards are available for analysis.

```python
plt.figure(figsize=(12,6))

plt.plot(df["Episode"], df["Reward"])

plt.title("Episode Reward During Training")

plt.xlabel("Episode")

plt.ylabel("Reward")

plt.grid(True)

plt.show()
```

## Observation

The graph illustrates how the reward changes over training episodes. An overall increasing trend indicates that the Reinforcement Learning agent is improving its pricing strategy over time.

```python
df["Moving_Average"] = df["Reward"].rolling(window=20).mean()
```

```python
plt.figure(figsize=(12,6))

plt.plot(df["Episode"], df["Reward"], label="Episode Reward")

plt.plot(
    df["Episode"],
    df["Moving_Average"],
    linewidth=3,
    label="Moving Average"
)

plt.title("Training Performance")

plt.xlabel("Episode")

plt.ylabel("Reward")

plt.legend()

plt.grid(True)

plt.show()
```

## Observation

The moving average smooths short-term reward fluctuations and clearly shows the overall learning trend of the agent.

```python
print("Maximum Reward :", df["Reward"].max())

print("Minimum Reward :", df["Reward"].min())

print("Average Reward :", round(df["Reward"].mean(),2))

print("Median Reward :", round(df["Reward"].median(),2))

print("Standard Deviation :", round(df["Reward"].std(),2))
```

## Observation

The performance metrics summarize the overall reward achieved during training.

These values help evaluate the consistency and effectiveness of the Reinforcement Learning agent.

```python
best_episode = df.loc[df["Reward"].idxmax()]

print(best_episode)
```

## Observation

The best-performing episode corresponds to the highest reward achieved during training.

```python
lowest_episode = df.loc[df["Reward"].idxmin()]

print(lowest_episode)
```

## Observation

The lowest reward generally occurs during the initial learning phase when the agent is still exploring the environment.
