# Reward vs Episode

## Objective

This visualization shows how the Reinforcement Learning (RL) agent improves during training.

As the number of training episodes increases, the agent learns better pricing strategies, resulting in higher cumulative rewards.

Since the RL model is still under development by the training team, simulated reward values are used for visualization.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Total training episodes

episodes = np.arange(1,101)

# Simulated rewards

np.random.seed(42)

reward = np.linspace(20,250,100) + np.random.normal(0,10,100)

reward = np.maximum(reward,0)

training_df = pd.DataFrame({
    "Episode":episodes,
    "Reward":reward
})

training_df.head()
```

```python
import matplotlib.pyplot as plt

# 1. Create the plot
plt.figure(figsize=(8, 5))
plt.plot(training_df["Episode"], training_df["Reward"], color="b", linewidth=2)

# 2. Add labels and title
plt.title("Reward vs Episode")
plt.xlabel("Episode")
plt.ylabel("Reward")
plt.grid(True)

# 3. Save the figure BEFORE plt.show()
plt.savefig("reward_vs_episode.png", bbox_inches="tight")
plt.show()
```

```python
average_reward = training_df["Reward"].mean()

print("Average Reward:",round(average_reward,2))
```

## Observation

The graph demonstrates that the reward generally increases as the number of training episodes increases.

Minor fluctuations occur due to the exploration performed by the Reinforcement Learning agent.

Overall, the upward trend indicates that the pricing agent is learning more profitable pricing strategies over time.

## Business Interpretation

Increasing rewards indicate improved pricing decisions.

As the Reinforcement Learning agent gains experience, it selects prices that maximize long-term revenue while maintaining customer demand.

This behavior supports revenue optimization in travel and hospitality systems.

# Dynamic Price vs Demand Level

## Objective

This visualization shows how the dynamically generated price affects customer demand.

The graph helps analyse whether increasing prices decreases customer demand, which is an important concept in Reinforcement Learning based Dynamic Pricing.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

plt.scatter(
    df["Dynamic_Price"],
    df["Demand_Level"],
    alpha=0.7
)

plt.title("Dynamic Price vs Demand Level")

plt.xlabel("Dynamic Price")

plt.ylabel("Demand Level")

plt.grid(True)
plt.savefig("price_vs_demand.png", dpi=300, bbox_inches="tight")
plt.show()
```

## Observation

The graph demonstrates the relationship between dynamic pricing and customer demand.

Generally, higher prices tend to reduce demand, while lower prices attract more bookings.

This relationship helps the Reinforcement Learning agent learn an optimal pricing policy.

# Revenue Comparison by Hotel Type

## Objective

Compare the average revenue generated by different hotel types.

```python
import matplotlib.pyplot as plt

revenue_by_hotel = df.groupby("Hotel_Type")["Revenue"].mean()

plt.figure(figsize=(8,6))

revenue_by_hotel.plot(kind="bar")

plt.title("Average Revenue by Hotel Type")

plt.xlabel("Hotel Type")

plt.ylabel("Average Revenue")

plt.grid(axis="y")
plt.savefig("revenue_comparison.png", dpi=300, bbox_inches="tight")
plt.show()
```

## Observation

The graph compares the average revenue generated by different hotel types.

It helps identify which hotel category contributes more revenue and provides useful insights for pricing strategy.

# Base Price vs Dynamic Price

## Objective

Compare the original base price with the dynamically generated price.

```python
average_prices = [
    df["Base_Price"].mean(),
    df["Dynamic_Price"].mean()
]

labels = [
    "Base Price",
    "Dynamic Price"
]

plt.figure(figsize=(7,5))

plt.bar(labels, average_prices)

plt.title("Average Base Price vs Dynamic Price")

plt.ylabel("Average Price")

for i, value in enumerate(average_prices):
    plt.text(i, value + 2, f"{value:.2f}", ha="center")
plt.savefig("price_comparison.png", dpi=300, bbox_inches="tight")
plt.show()
```
