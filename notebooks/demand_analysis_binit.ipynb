```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("cleaned_hotel_bookings.csv")

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
monthly_bookings = df["arrival_date_month"].value_counts()

monthly_bookings.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Monthly Booking Demand")
plt.xlabel("Month")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=45)
plt.show()
```

```python
hotel_type = df["hotel"].value_counts()

hotel_type.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6,6)
)

plt.title("Hotel Type Distribution")
plt.ylabel("")
plt.show()
```

```python
cancellations = df["is_canceled"].value_counts()

cancellations.plot(
    kind="bar"
)

plt.title("Booking Cancellations")
plt.xlabel("Cancelled")
plt.ylabel("Number of Bookings")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["adr"], bins=30)

plt.title("Average Daily Rate Distribution")
plt.xlabel("ADR")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["lead_time"], bins=30)

plt.title("Lead Time Distribution")
plt.xlabel("Lead Time (Days)")
plt.ylabel("Frequency")

plt.show()
```

```python
customer = df["customer_type"].value_counts()

customer.plot(kind="bar")

plt.title("Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Bookings")

plt.show()
```

```python
segment = df["market_segment"].value_counts()

segment.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Market Segment Distribution")
plt.xlabel("Market Segment")
plt.ylabel("Bookings")
plt.xticks(rotation=45)

plt.show()
```

```python
status = df["reservation_status"].value_counts()

status.plot(kind="bar")

plt.title("Reservation Status")
plt.xlabel("Status")
plt.ylabel("Bookings")

plt.show()
```

```python
import matplotlib.pyplot as plt

numeric_df = df.select_dtypes(include=['number'])

plt.figure(figsize=(10,8))

plt.imshow(numeric_df.corr(), cmap="coolwarm")

plt.colorbar()

plt.xticks(range(len(numeric_df.columns)), numeric_df.columns, rotation=90)
plt.yticks(range(len(numeric_df.columns)), numeric_df.columns)

plt.title("Correlation Heatmap")

plt.show()
```

Summary

The demand analysis identified important booking trends such as seasonal demand, hotel preferences, cancellation patterns, pricing distribution, and customer segments. These insights provide useful information for designing a Reinforcement Learning-based dynamic pricing model that can adjust prices according to customer demand.
