## Dynamic Pricing Visualizations

### Objective

To create visualizations explaining dynamic pricing and reinforcement learning concepts for the Travel and Hospitality project.

```python
import matplotlib.pyplot as plt
price=[50,75,100,125,150,175,200]
demand=[100,90,80,65,50,30,15]
plt.figure(figsize=(8,5))
plt.plot(price,demand,marker='o',linewidth=2)
plt.title("Demand vs Price")
plt.xlabel("Price ($)")
plt.ylabel("Demand")
plt.grid(True)
plt.savefig("demand_vs_price.png",dpi=300)
plt.show()
```

### Observation

Customer demand decreases as ticket prices increase.

This illustrates the negative relationship between price and demand.

```python
price=[50,75,100,125,150,175,200]
revenue=[5000,6800,7900,8200,7600,6000,3500]
plt.figure(figsize=(8,5))
plt.plot(price,revenue,marker='o',linewidth=2)
plt.title("Revenue vs Price")
plt.xlabel("Price ($)")
plt.ylabel("Revenue ($)")
plt.grid(True)
plt.savefig("revenue_vs_price.png",dpi=300)
plt.show()
```

Revenue reaches its highest value at an optimal ticket price.

Increasing prices beyond this point reduces revenue because demand falls.

```python
price=[50,75,100,125,150,175,200]
booking=[0.95,0.87,0.74,0.58,0.42,0.23,0.08]
plt.figure(figsize=(8,5))
plt.plot(price,booking,marker='o',linewidth=2)
plt.title("Booking Probability vs Price")
plt.xlabel("Price ($)")
plt.ylabel("Booking Probability")
plt.grid(True)
plt.savefig("booking_probability.png",dpi=300)
plt.show()
```

Higher prices reduce the probability of customers making a booking.
