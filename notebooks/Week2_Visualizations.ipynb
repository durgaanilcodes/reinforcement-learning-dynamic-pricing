## 1. Introduction

Dynamic pricing is a pricing strategy where the price of a product or service changes based on customer demand, inventory availability, time, seasonality, and market conditions.

In industries such as airlines and hotels, prices are continuously adjusted to maximize revenue while ensuring that limited inventory (such as seats or rooms) is sold efficiently.

The following visualizations illustrate the key relationships used in dynamic pricing systems.

## 2. Demand vs Price Visualization

### Objective

This graph illustrates the relationship between ticket price and customer demand.

As the ticket price increases, the number of customers willing to purchase generally decreases.

```python
import matplotlib.pyplot as plt
price=[50,75,100,125,150,175,200]
demand=[100,90,80,65,50,30,15]
plt.figure(figsize=(8,5))
plt.plot(price,demand,marker='o',linewidth=2)
plt.title("Demand vs Price")
plt.xlabel("Price ($)")
plt.ylabel("Demand")
plt.grid(True)
plt.savefig("demand_vs_price.png",dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

Customer demand decreases as ticket prices increase.

This illustrates the negative relationship between price and demand.

## 3. Revenue vs Price Visualization

### Objective

Revenue does not always increase with higher prices.

Although increasing prices may improve profit per booking, excessive prices reduce demand, which ultimately decreases overall revenue.

This graph demonstrates the concept of an optimal ticket price.

```python
price=[50,75,100,125,150,175,200]
revenue=[5000,6800,7900,8200,7600,6000,3500]
plt.figure(figsize=(8,5))
plt.plot(price,revenue,marker='o',linewidth=2)
plt.title("Revenue vs Price")
plt.xlabel("Price ($)")
plt.ylabel("Revenue ($)")
plt.grid(True)
plt.savefig("revenue_vs_price.png",dpi=300)
plt.show()
```

Revenue reaches its highest value at an optimal ticket price.

Increasing prices beyond this point reduces revenue because demand falls.

## Summary

The two visualizations demonstrate the fundamental relationships used in dynamic pricing:

- Higher prices generally reduce customer demand.
- Revenue is maximized at an optimal price rather than the highest possible price.
- These concepts motivate the use of Reinforcement Learning, where an agent learns pricing strategies that maximize long-term revenue.

## 4. Booking Probability vs Price

### Objective

Booking probability represents the likelihood that a customer will purchase a ticket at a given price.

Generally, as ticket prices increase, customers become less likely to complete a booking.

```python
price=[50,75,100,125,150,175,200]
booking=[0.95,0.87,0.74,0.58,0.42,0.23,0.08]
plt.figure(figsize=(8,5))
plt.plot(price,booking,marker='o',linewidth=2)
plt.title("Booking Probability vs Price")
plt.xlabel("Price ($)")
plt.ylabel("Booking Probability")
plt.grid(True)
plt.savefig("booking_probability.png",dpi=300)
plt.show()
```

Higher prices reduce the probability of customers making a booking.

## 5. Demand Heatmap

### Objective

A heatmap provides a visual representation of demand intensity across different ticket prices.

Darker colours indicate higher customer demand, while lighter colours represent lower demand.

```python
import pandas as pd
```

```python
import numpy as np

prices = [50,75,100,125,150,175,200]

days = [
    "30 Days",
    "25 Days",
    "20 Days",
    "15 Days",
    "10 Days",
    "5 Days",
    "Departure"
]

demand = np.array([
    [98,94,90,82,70,55,40],
    [92,88,82,74,60,46,30],
    [84,80,74,66,52,38,24],
    [74,70,64,56,44,30,18],
    [60,56,50,42,30,22,12],
    [42,38,34,28,20,14,8],
    [24,20,18,14,10,6,2]
])

plt.figure(figsize=(8,6))

plt.imshow(demand,
           cmap="YlOrRd",
           aspect="auto")

plt.colorbar(label="Demand")

plt.xticks(range(len(days)),days)

plt.yticks(range(len(prices)),prices)

plt.xlabel("Days Before Departure")

plt.ylabel("Ticket Price (£)")

plt.title("Demand Heatmap")

plt.savefig("demand_heatmap.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

### Observation

The heatmap shows that customer demand is generally higher at lower ticket prices and decreases as prices increase.

It also demonstrates how demand changes as the departure date approaches, illustrating why dynamic pricing strategies continuously adjust prices over time.

## 6. Dynamic Price Trend

### Objective

This visualization illustrates how ticket prices may change as the departure date approaches.

Dynamic pricing systems adjust prices based on demand, inventory, and remaining time.

```python
days = [30,25,20,15,10,5,1]

price = [80,90,105,120,145,165,185]

plt.figure(figsize=(8,5))

plt.plot(days,
         price,
         marker='o',
         linewidth=2)

plt.gca().invert_xaxis()

plt.title("Dynamic Ticket Price Trend")

plt.xlabel("Days Before Departure")

plt.ylabel("Ticket Price (£)")

plt.grid(True)

plt.savefig("price_trend.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

### Observation

The graph demonstrates a typical pricing strategy where ticket prices increase as the departure date approaches and seat availability decreases.

In real-world systems, Reinforcement Learning can dynamically adjust these prices by considering demand, inventory, and expected future revenue rather than following a fixed pricing rule.

# 7. Reinforcement Learning Workflow

The Reinforcement Learning workflow illustrates how an intelligent pricing agent interacts with the market environment.

The agent observes the current state, selects an appropriate ticket price, receives a reward based on the generated revenue, and updates its knowledge through continuous interaction.

This learning process enables the agent to discover pricing strategies that maximize long-term revenue instead of relying on fixed pricing rules.

# 8. Q-Learning Workflow

Q-Learning is a value-based Reinforcement Learning algorithm.

For each state, the agent evaluates all possible pricing actions and gradually learns which action generates the highest long-term reward.

The Q-Table stores the expected future reward for every state-action pair, allowing the agent to improve its pricing policy through repeated interactions with the environment.
