## Introduction

Reinforcement Learning (RL) is a branch of Machine Learning where an intelligent agent learns by interacting with an environment. The agent performs actions, receives rewards or penalties, and gradually learns the best strategy to maximize cumulative rewards.

This documentation summarizes the fundamental concepts of Reinforcement Learning that are required for implementing a Dynamic Pricing system.

## What is Reinforcement Learning?

Reinforcement Learning is a learning technique where an agent continuously interacts with an environment.

The agent performs actions.

The environment provides rewards.

The agent learns from those rewards.

The goal is to maximize long-term rewards.

## Basic Components

### Agent

The decision maker.

### Environment

The system in which the agent operates.

### State

Current situation of the environment.

### Action

Decision taken by the agent.

### Reward

Feedback received after an action.

### Policy

Strategy used to choose actions.

### Episode

One complete learning cycle.

# Markov Decision Process (MDP)

MDP is a mathematical framework used in Reinforcement Learning.

Components of MDP:

State (S)

Action (A)

Reward (R)

Policy (π)

Transition Probability (P)

The future depends only on the current state and action.
