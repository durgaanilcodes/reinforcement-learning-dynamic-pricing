# Revenue by Hotel Type

This chart compares the total revenue generated by different hotel types.

```python
plt.figure(figsize=(8,5))
revenue = df.groupby("Hotel_Type")["Revenue"].sum()
plt.bar(revenue.index, revenue.values)
plt.title("Revenue by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Total Revenue")
plt.grid(axis='y')
plt.savefig("revenue_by_hoteltype.png", dpi=300,bbox_inches="tight")
plt.show()
```

# Revenue by Season

This visualization shows seasonal revenue variation.

```python
plt.figure(figsize=(8,5))
season_rev = df.groupby("Season")["Revenue"].sum()
plt.bar(season_rev.index, season_rev.values)
plt.title("Revenue by Season")
plt.xlabel("Season")
plt.ylabel("Revenue")
plt.grid(axis='y')
plt.savefig("revenue_by_season.png", dpi=300,bbox_inches="tight")
plt.show()
```

# Average Dynamic Price by Hotel Type

```python
plt.figure(figsize=(8,5))
price = df.groupby("Hotel_Type")["Dynamic_Price"].mean()
plt.bar(price.index, price.values)
plt.title("Average Dynamic Price")
plt.xlabel("Hotel Type")
plt.ylabel("Average Price")
plt.grid(axis='y')
plt.savefig("average_dynamic_price.png", dpi=300,bbox_inches="tight")
plt.show()
```

# Booking Probability Distribution

```python
plt.figure(figsize=(8,5))
plt.hist(df["Booking_Probability"], bins=20)
plt.title("Booking Probability Distribution")
plt.xlabel("Booking Probability")
plt.ylabel("Frequency")
plt.grid()
plt.savefig("booking_probability.png", dpi=300,bbox_inches="tight")
plt.show()
```
