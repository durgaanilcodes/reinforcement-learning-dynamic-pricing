# Revenue by Hotel Type

This chart compares the total revenue generated by different hotel types.

```python
plt.figure(figsize=(8,5))
revenue = df.groupby("Hotel_Type")["Revenue"].sum()
plt.bar(revenue.index, revenue.values)
plt.title("Revenue by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Total Revenue")
plt.grid(axis='y')
plt.savefig("revenue_by_hoteltype.png", dpi=300,bbox_inches="tight")
plt.show()
```

# Revenue by Season

This visualization shows seasonal revenue variation.

```python
plt.figure(figsize=(8,5))
season_rev = df.groupby("Season")["Revenue"].sum()
plt.bar(season_rev.index, season_rev.values)
plt.title("Revenue by Season")
plt.xlabel("Season")
plt.ylabel("Revenue")
plt.grid(axis='y')
plt.savefig("revenue_by_season.png", dpi=300,bbox_inches="tight")
plt.show()
```

# Pricing Strategy Before vs After RL Optimization

In this section, we compare the traditional pricing strategy (Base Price) with the Reinforcement Learning optimized pricing (Dynamic Price).

These comparisons help evaluate whether the RL agent generated better pricing decisions and improved revenue.

```python
plt.figure(figsize=(10,6))
plt.plot(df["Base_Price"], label="Base Price")
plt.plot(df["Dynamic_Price"], label="Dynamic Price")
plt.title("Base Price vs Dynamic Price")
plt.xlabel("Booking Records")
plt.ylabel("Price")
plt.legend()
plt.grid()
plt.savefig("base_vs_dynamic_price.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
df["Revenue_Before_RL"] = (
    df["Base_Price"] *
    df["Booking_Probability"]
)
```

```python
df["Revenue"]
```

```python
comparison = pd.DataFrame({
    "Before RL":[df["Revenue_Before_RL"].sum()],
    "After RL":[df["Revenue"].sum()]
})

comparison
```

```python
plt.figure(figsize=(7,5))

plt.bar(
    comparison.columns,
    comparison.iloc[0]
)

plt.title("Revenue Before vs After RL")

plt.ylabel("Revenue")

plt.grid(axis="y")

plt.show()
```

```python
avg_price = pd.DataFrame({
    "Pricing":[
        "Base Price",
        "Dynamic Price"
    ],
    "Average":[
        df["Base_Price"].mean(),
        df["Dynamic_Price"].mean()
    ]

})
avg_price
```

```python
plt.figure(figsize=(7,5))
plt.bar(
    avg_price["Pricing"],
    avg_price["Average"]
)
plt.title("Average Price Comparison")
plt.ylabel("Average Price")
plt.grid(axis="y")
plt.show()
```
