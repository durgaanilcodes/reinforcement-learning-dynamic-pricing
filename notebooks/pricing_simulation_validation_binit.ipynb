```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("demand_scenarios.csv")

print("Dataset loaded successfully!")
```

```python
df.head()
```

## Observation

The demand scenario dataset contains pricing information, booking probability, and expected revenue for different demand conditions.

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
plt.figure(figsize=(8,5))

x = range(len(df))

plt.bar(x, df["Base_Price"], width=0.4, label="Base Price")
plt.bar([i + 0.4 for i in x], df["Dynamic_Price"], width=0.4, label="Dynamic Price")

plt.xticks([i + 0.2 for i in x], df["Scenario"], rotation=20)

plt.title("Base Price vs Dynamic Price")

plt.xlabel("Demand Scenario")
plt.ylabel("Price")

plt.legend()

plt.show()
```

## Observation

Dynamic prices increase during high-demand scenarios and decrease during low-demand scenarios, indicating that the pricing logic responds appropriately to changes in demand.

```python
plt.figure(figsize=(8,5))

plt.plot(
    df["Scenario"],
    df["Expected_Revenue"],
    marker="o",
    linewidth=3
)

plt.title("Expected Revenue Across Demand Scenarios")

plt.xlabel("Scenario")

plt.ylabel("Revenue")

plt.grid(True)

plt.show()
```

## Observation

Revenue is highest during Peak Season and High Demand scenarios due to increased booking probability and higher prices.

```python
plt.figure(figsize=(8,5))

plt.bar(
    df["Scenario"],
    df["Booking_Probability"]
)

plt.title("Booking Probability Comparison")

plt.xlabel("Scenario")

plt.ylabel("Probability")

plt.xticks(rotation=20)

plt.show()
```

## Observation

Booking probability increases with demand, which matches expected customer behavior.
