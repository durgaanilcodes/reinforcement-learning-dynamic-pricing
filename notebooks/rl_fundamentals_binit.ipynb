### Introduction

Reinforcement Learning (RL) is a branch of Machine Learning in which an intelligent agent learns by interacting with an environment. The agent performs actions, receives rewards or penalties, and gradually learns the best strategy to maximize the total reward.

# Basic Concepts of Reinforcement Learning

### 1. Agent

An agent is the learner or decision-maker in Reinforcement Learning.

Example:
A pricing AI that decides the price of a hotel room.

### 2. Environment

The environment is everything with which the agent interacts.

Example:
The hotel booking system where customers book rooms.

### 3. State

A state represents the current situation of the environment.

Example:

Rooms Available = 20

Days Remaining = 15

State = (20,15)

### 4. Action

An action is the decision made by the agent.

Examples:

- Increase room price
- Decrease room price
- Keep price unchanged

### 5. Reward

Reward is the feedback received after taking an action.

Example:

Room sold

Reward = ₹5000

Room not sold

Reward = ₹0

### 6. Policy

A policy is the strategy that tells the agent which action to take in every state.

### 7. Episode

An episode is one complete cycle of interaction between the agent and the environment.

Example:

Selling hotel rooms from Day 30 until Day 1.

### 8. Exploration

Exploration means trying different actions to discover better strategies.

### 9. Exploitation

Exploitation means selecting the best-known action to maximize reward.
